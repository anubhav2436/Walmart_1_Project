8.Categorize sales into 3 group MORNING, AFTERNOON, EVENING
Find out each of the shift and number of invoices

```python
df.drop('mean1',axis=1,inplace=True)
```

```python
df.rename(columns=  {'Total':'revenue'},inplace=True)
```

```python
# df['time']=pd.to_datetime(df['time'])
df.time.dtype
```

```python


def timing_zone2(t):
  hours = t.hour
  if hours < 12:
    return 'morning'
  elif 12 <= hours < 18:
    return 'afternoon'
  else:
    return 'night'
```

```python
df['time_zone']=df['time'].apply(timing_zone2)
```

```python
df.groupby('time_zone')['invoice_id'].count().reset_index(name='count_of_invoice')
```

9. Identify 5 branch with highest decrese ratio in
-- revevenue compare to last year(current year 2023 and last year 2022)
-- rdr == last_rev-cr_rev/ls_rev*100

```python
df['date']=pd.to_datetime(df['date'],format='mixed')
df['year']=df['date'].dt.year
```

```python
df2=df.groupby(['branch','year'])['revenue'].sum().reset_index(name='revenue_sum')
```

```python
df_2022=df2[df2.year.isin([2022])]
```

```python
df_2023=df2[df2.year.isin([2023])]
```

```python
df3=pd.merge(df_2022,df_2023,on='branch',how='inner')
```

```python
df3['ratio']=(df3['revenue_sum_x']-df3['revenue_sum_y'])/df3['revenue_sum_x']*100
```

```python
df3[['branch','ratio']].sort_values(by='ratio',ascending=False).head(5)
```
